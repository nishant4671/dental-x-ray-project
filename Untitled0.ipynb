```python
import os
import numpy as np

def convert_box_to_seg(box_line, img_w, img_h):
    # box_line: "class cx cy w h" (normalized)
    parts = box_line.strip().split()
    cls = parts[0]
    cx, cy, w, h = map(float, parts[1:])

    # Convert center+size to corner coordinates (normalized)
    x1 = cx - w/2
    y1 = cy - h/2
    x2 = cx + w/2
    y2 = cy + h/2

    # Polygon: x1 y1 x2 y1 x2 y2 x1 y2 (clockwise)
    return f"{cls} {x1:.6f} {y1:.6f} {x2:.6f} {y1:.6f} {x2:.6f} {y2:.6f} {x1:.6f} {y2:.6f}"

# Convert all labels in train/val/test
for split in ['train', 'val', 'test']:
    label_dir = f'cavity_data/cavity_dataset/{split}/labels'
    if not os.path.exists(label_dir):
        continue
    for file in os.listdir(label_dir):
        if file.endswith('.txt'):
            with open(os.path.join(label_dir, file), 'r') as f:
                lines = f.readlines()
            # Convert each line
            new_lines = []
            for line in lines:
                new_lines.append(convert_box_to_seg(line, 1, 1))
            with open(os.path.join(label_dir, file), 'w') as f:
                f.write('\n'.join(new_lines))

print("Conversion done! Labels now have 8 coordinates (polygon).")
```

```python
import os

def convert_box_to_seg(line):
    parts = line.strip().split()
    if len(parts) < 5:
        return line  # skip empty/malformed
    cls = parts[0]
    cx, cy, w, h = map(float, parts[1:])
    x1 = cx - w/2
    y1 = cy - h/2
    x2 = cx + w/2
    y2 = cy + h/2
    return f"{cls} {x1:.6f} {y1:.6f} {x2:.6f} {y1:.6f} {x2:.6f} {y2:.6f} {x1:.6f} {y2:.6f}"

for split in ['train', 'val', 'test']:
    label_dir = f'cavity_data/cavity_dataset/{split}/labels'
    if not os.path.exists(label_dir):
        continue
    for file in os.listdir(label_dir):
        if file.endswith('.txt'):
            path = os.path.join(label_dir, file)
            with open(path, 'r') as f:
                lines = f.readlines()
            new_lines = [convert_box_to_seg(l) for l in lines]
            with open(path, 'w') as f:
                f.write('\n'.join(new_lines))
print("✅ Conversion done!")
```

```python
from ultralytics import YOLO
model = YOLO("yolov8n-seg.pt")
model.train(
    data="cavity_data/cavity_dataset/dataset.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    workers=4,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=15,
    flipud=0.5,
    fliplr=0.5,
    project="cavity_training",
    name="exp1",
    exist_ok=True,
    save=True
)
```
